```python
import json, torch
from transformers import AutoTokenizer, AutoModel
from typing import List
import numpy as np

PATHS = {
    "embed_model": "/home/aiteam/Desktop/rag_pipelines/models/colnomic-embed-multimodal-3b",
    # "pdf": "/home/aiteam/Desktop/rag_pipelines/src/multimodal_rag/data/file_1.pdf"
}


tokE = AutoTokenizer.from_pretrained(PATHS["embed_model"], local_files_only=True)
modE = AutoModel.from_pretrained(PATHS["embed_model"], local_files_only=True, torch_dtype=torch.float32)
modE.eval()

torch.no_grad()
def encode_text(texts: List[str]) -> np.ndarray:
    out_vecs=[]
    for batch_start in range(0, len(texts), 4):
        batch = texts[batch_start:batch_start+4]
        enc = tokE(batch, padding=True, truncation=True, return_tensors="pt", max_length=512)
        last = modE(**enc).last_hidden_state   # [B, T, H]
        mask = enc["attention_mask"].unsqueeze(-1)          # [B, T, 1]
        summed = (last * mask).sum(dim=1)                   # [B, H]
        denom = mask.sum(dim=1).clamp(min=1e-9)             # [B, 1]
        mean = (summed / denom).cpu().numpy()               # [B, H]
        out_vecs.append(mean)
    return np.vstack(out_vecs).astype("float32")

DOC_EMB = encode_text(chunks)
print("✅ Document embeddings generated!")
print("Number of chunks:", len(chunks))
print("Embeddings shape:", DOC_EMB.shape)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from typing import List, Tuple
import numpy as np
import torch

PATHS = {
    "reranker": "/home/aiteam/Desktop/rag_pipelines/models/bge-reranker-v2-m3"
}
tokR = AutoTokenizer.from_pretrained(PATHS["reranker"], local_files_only=True)
modR = AutoModelForSequenceClassification.from_pretrained(PATHS["reranker"], local_files_only=True)
modR.eval()


def initial_retrieve(query: str, topk: int = 12, embeddings: np.ndarray = None):
    if embeddings is None:
        embeddings = DOC_EMB
    qv = encode_text([query])[0]
    # vectorized cosine (faster/cleaner)
    emb_norm = embeddings / (np.linalg.norm(embeddings, axis=1, keepdims=True) + 1e-12)
    qv_norm = qv / (np.linalg.norm(qv) + 1e-12)
    sims = emb_norm @ qv_norm
    idx = np.argsort(sims)[::-1][:topk]
    return [(int(i), float(sims[i])) for i in idx]

# def initial_retrieve(query:str, topk=12)->List[Tuple[int,float]]:
#     qv = encode_text([query])[0]
#     sims = [cosine_sim(qv, v) for v in DOC_EMB]
#     idx = np.argsort(sims)[::-1][:topk]
#     return [(int(i), float(sims[i])) for i in idx]

@torch.no_grad()
def bge_rerank(query:str, cand_idx:List[int], topk=5)->List[int]:
    pairs = [(query, chunks[i]) for i in cand_idx]
    enc = tokR([p[0] for p in pairs],[p[1] for p in pairs],
               truncation=True, padding=True, return_tensors="pt", max_length=512)
    scores = modR(**enc).logits.squeeze(-1)   # larger → more relevant
    order = torch.argsort(scores, descending=True).tolist()
    ranked = [cand_idx[i] for i in order[:topk]]
    return ranked
```

```python
# 5) Robust Kokoro TTS loader (handles multiple package layouts)

import importlib, types

PATHS = {
    "kokoro_onnx": "/home/aiteam/Desktop/rag_pipelines/models/kokoro/kokoro-v1.0.int8.onnx",
    "kokoro_voices": "/home/aiteam/Desktop/rag_pipelines/models/kokoro/voices-v1.0.bin",
    "wav_out": "/home/aiteam/Desktop/rag_pipelines/src/multimodal_rag/voice_notebook/output.wav"
}

def _try_import(path, member_names=("TTS","Kokoro","KokoroTTS")):
    """
    Try to import a module and return (module, found_member_name or None).
    """
    try:
        m = importlib.import_module(path)
    except Exception:
        return None, None
    for n in member_names:
        if hasattr(m, n):
            return m, n
    return m, None

def load_kokoro(model_path: str, voices_path: str):
    """
    Returns (ktts_obj, impl_string). Tries several known variants.
    """
    tried = []

    # 1) kokoro_tts (PyPI: kokoro-tts)
    m, n = _try_import("kokoro_tts")
    if m and n:
        K = getattr(m, n)
        try:
            return K(model_path=model_path, voices_path=voices_path), f"kokoro_tts.{n}"
        except TypeError:
            # different arg names
            try:
                return K(model=model_path, voices=voices_path), f"kokoro_tts.{n}"
            except Exception as e:
                tried.append(("kokoro_tts", repr(e)))

    # 2) kokoro_onnx at root
    m, n = _try_import("kokoro_onnx")
    if m and n:
        K = getattr(m, n)
        for kwargs in (
            dict(model_path=model_path, voices_path=voices_path),
            dict(model=model_path, voices=voices_path),
        ):
            try:
                return K(**kwargs), f"kokoro_onnx.{n}"
            except Exception as e:
                tried.append(("kokoro_onnx", repr(e)))

    # 3) kokoro_onnx submodules that are commonly used
    for sub in ("tts", "inference", "infer", "api"):
        m, n = _try_import(f"kokoro_onnx.{sub}")
        if m and n:
            K = getattr(m, n)
            for kwargs in (
                dict(model_path=model_path, voices_path=voices_path),
                dict(model=model_path, voices=voices_path),
            ):
                try:
                    return K(**kwargs), f"kokoro_onnx.{sub}.{n}"
                except Exception as e:
                    tried.append((f"kokoro_onnx.{sub}", repr(e)))

    raise ImportError(
        "Could not construct a Kokoro TTS object from installed packages.\n"
        "Tried variants: " + "; ".join([f"{a} -> {b}" for a,b in tried])
    )

# ---- create a global TTS instance ----
ktts, kokoro_impl = load_kokoro(PATHS["kokoro_onnx"], PATHS["kokoro_voices"])
print("Kokoro loaded via:", kokoro_impl)

def speak_to_wav(text: str, out_path: str = PATHS["wav_out"]):
    """
    Call the available synth method and save a mono WAV. Handles bytes or numpy output.
    """
    # find a callable: tts / synthesize / synth / __call__
    synth_fn = None
    for name in ("tts", "synthesize", "synth", "__call__"):
        if hasattr(ktts, name):
            synth_fn = getattr(ktts, name)
            break
    if synth_fn is None:
        raise RuntimeError("Kokoro object has no tts/synthesize/synth/__call__ method.")

    audio = synth_fn(text)

    import wave, numpy as np
    # If bytes-like, assume already WAV
    if isinstance(audio, (bytes, bytearray)):
        with open(out_path, "wb") as f:
            f.write(audio)
        return out_path

    # Otherwise assume float numpy in [-1, 1] and use a default/fetched sample rate
    sr = getattr(ktts, "sample_rate", 22050)
    audio_i16 = (np.clip(np.array(audio), -1.0, 1.0) * 32767).astype(np.int16)
    with wave.open(out_path, "wb") as wf:
        wf.setnchannels(1)
        wf.setsampwidth(2)
        wf.setframerate(sr)
        wf.writeframes(audio_i16.tobytes())
    return out_path
```

```python
# 6) Orchestration with selectable voice (text → RAG → TTS)

def rag_answer_with_voice(
    query: str,
    k_init: int = 12,
    k_rerank: int = 5,
    voice: str | None = None,   # e.g., "af_sarah", "am_adam", etc.
    speed: float = 1.0,
    lang: str = "en-us",
    out_wav: str | None = None
):
    """
    Runs: embed → retrieve → rerank → generate → Kokoro TTS.
    - voice: pick from VOICE_NAMES; if None, uses CURRENT_VOICE.
    - returns answer text, wav path, and the top chunk ids for transparency.
    """
    if out_wav is None:
        out_wav = PATHS["wav_out"]

    # 1) initial retrieve from precomputed DOC_EMB
    cand = initial_retrieve(query, topk=k_init)
    cand_idx = [i for i, _ in cand]

    # 2) rerank with BGE cross-encoder
    top_idx = bge_rerank(query, cand_idx, topk=k_rerank)

    # 3) build short context from top chunks
    context = "\n---\n".join(chunks[i] for i in top_idx)

    # 4) generate with local LLM
    answer = generate_answer(query, context)

    # 5) synthesize with Kokoro (selected voice)
    wav_path = speak_to_wav(answer, out_path=out_wav, voice=voice, speed=speed, lang=lang)

    # return both modalities
    return {
        "answer_text": answer,
        "audio_file": wav_path,
        "ctx_ids": top_idx,
        "voice_used": voice if voice is not None else globals().get("CURRENT_VOICE", None)
    }

# --- quick smoke tests ---

# A) use the global default voice (CURRENT_VOICE)
res_default = rag_answer_with_voice("Give me a concise summary of the main points in the PDF.")
print("Default voice:", res_default["voice_used"])
print("Top chunk ids:", res_default["ctx_ids"])
print("Answer:", res_default["answer_text"][:300], "...")
print("Audio file:", res_default["audio_file"])

# B) override voice just for this call
res_adam = rag_answer_with_voice(
    "List two key takeaways from the document.",
    voice="am_adam",   # must exist in VOICE_NAMES
    speed=1.0
)
print("\nOverride voice:", res_adam["voice_used"])
print("Top chunk ids:", res_adam["ctx_ids"])
print("Answer:", res_adam["answer_text"][:300], "...")
print("Audio file:", res_adam["audio_file"])
```
